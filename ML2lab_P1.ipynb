```python
df = pd.read_csv("/content/StudentD.csv")
print(df)
df['StudyHours'].fillna(df['StudyHours'].mode(),inplace=True)
df['Attendance'].fillna(df['Attendance'].mode(),inplace=True)
df['Marks'].fillna(df['Marks'].mode(),inplace=True)
```
